# SCRAPE COMMENTS

```python
import googleapiclient.discovery
import googleapiclient.errors
import pandas as pd

api_service_name = "youtube"
api_version = "v3"
DEVELOPER_KEY = "AIzaSyBSTSk5V_3M-HfXV-FP926i4ekh3maNHlI"

youtube = googleapiclient.discovery.build(
    api_service_name, api_version, developerKey=DEVELOPER_KEY)

video_id = "R9mN-d2CDUU"  # Replace with the YouTube video ID you want to fetch comments from
youtube_id= "https://www.youtube.com/watch?v=R9mN-d2CDUU"
max_results = 200  # Adjust as needed

# Define the keywords you want to search for
keywords = ["Helpful video", "paleo", '18','19','20','21','22','23','24','implemented','years','months','days','old' , 'active', 'workout', 'exercise', 'problem', 'craving', 'i am on', 'success', 'i']

comments = []

next_page_token = None

while True:
    request = youtube.commentThreads().list(
        part="snippet",
        videoId=video_id,
        maxResults=max_results,
        pageToken=next_page_token if next_page_token else ''
    )
    response = request.execute()

    for item in response['items']:
        comment = item['snippet']['topLevelComment']['snippet']
        text_display = comment['textDisplay']

        # Check if any of the keywords are present in the comment
        if any(keyword in text_display for keyword in keywords):
            comments.append([
                comment['authorDisplayName'],
                comment['publishedAt'],
                comment['updatedAt'],
                comment['likeCount'],
                text_display,
                youtube_id
            ])

    # Check if there are more pages of comments
    next_page_token = response.get('nextPageToken')

    if not next_page_token:
        break

df = pd.DataFrame(comments, columns=['author', 'published_at', 'updated_at', 'like_count', 'text', 'video'])

# Save the data frame to a CSV file
df.to_csv('paleoDiet3.csv', index=False)
df
```
